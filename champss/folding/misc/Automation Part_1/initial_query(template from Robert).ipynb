```python
from sps_pipeline.candidate_viewer import CandidateViewerQuery
import os  # if you need file checks later

# Database configuration
db_config = {
    'host': 'sps-archiver1',
    'user': 'automation',
    'port': 3306,
    'password': '',
    'database': 'champss'
}

# query = CandidateViewerQuery(survey="stackcands", db_config=db_config)
# candidates = query.get_metadata(folder="stack_0")

with CandidateViewerQuery(survey="stackcands", db_config=db_config) as query:
    # Get candidates rated as "NEW CANDIDATE" with metadata
    new_candidates = query.get_ratings(
        folder="stack_0",
        classification='NEW CANDIDATE',
        with_metadata=True
    )

    print(f"Found {len(new_candidates)} new candidates in stack_0\n")

    for cand in new_candidates:
        # print(cand)
        # Extract metadata
        metadata = cand['metadata']

        # Get values
        cand_name = metadata['file']
        dm = metadata['best_dm']
        p_bary_ms = float(metadata['p_bary'])

        # Calculate f0 from period (p_bary is in milliseconds)
        f0 = 1000.0 / p_bary_ms if p_bary_ms != 0 else 0

        # Candidate file path
        file_path = metadata['input_file']
        result = cand['result']

        print(f"Candidate: {cand_name}")
        print(f"  DM: {dm} pc/cm³")
        print(f"  f0: {f0:.6f} Hz")
        print(f"  File: {file_path}")
        print(f"  Result: {result}")
```
